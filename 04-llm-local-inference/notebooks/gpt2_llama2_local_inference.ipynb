```python
from transformers import GPT2Tokenizer, GPT2Model, GPT2LMHeadModel, TFGPT2LMHeadModel

#carga del modelo
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = TFGPT2LMHeadModel.from_pretrained("gpt2")
```

```python
model
```

```python
#generar texto con el modelo
input_text = "what is the capital of Colombia?"
input_ids = tokenizer.encode(input_text, return_tensors ='tf')

output =  model.generate(input_ids, max_length = 50, num_return_sequences =1, temperature =1.0, num_beams=5, top_k = 50, top_p = 0.95 )

```

```python
#decodificar las secuencias

generated_text = tokenizer.decode(output[0], skip_special_tokens = True)
print(generated_text)
```

```python
# Carga del modelo
model_name_or_path = "TheBloke/Llama-2-13B-chat-GGML"
model_basename = "llama-2-13b-chat.ggmlv3.q5_1.bin" # the model is in bin format
```

```python
from huggingface_hub import hf_hub_download
from llama_cpp import Llama
```

```python
model_path = hf_hub_download(repo_id=model_name_or_path, filename=model_basename)
```

```python

# GPU
lcpp_llm = None
lcpp_llm = Llama(
    model_path=model_path,
    n_threads=2, # CPU cores
    n_batch=512, # Should be between 1 and n_ctx, consider the amount of VRAM in your GPU.
    n_gpu_layers=32 # Change this value based on your model and your GPU VRAM pool.
    )

# See the number of layers in GPU
lcpp_llm.params.n_gpu_layers
```

```python
# Prompts
prompt = "Cuales son los efectos de la frecuencia de los impulsos utilizada en los entrenamientos con Electroestimulación Muscular (EMS) sobre el cuerpo humano?"

prompt_template=f'''SYSTEM: Eres  coy trainer, encargado de brindar los ejercicios, repeticiones, duración y paramétros de electro-estimulación (ems) para una sesión de entrenamiento. Todas las respuestas en español, no solicites más información y solo responde la información solicitada

USER: {prompt}

ASSISTANT:
'''
```

```python
response=lcpp_llm(prompt=prompt_template, max_tokens=1000, temperature=1, top_p=1,
                  repeat_penalty=1.3, top_k=300,
                  echo=True)
```

```python
print(response["choices"][0]["text"])
```

```python
#Dividir el texto en partes
full_response = response["choices"][0]["text"]
parts = full_response.split("USER:")
if len(parts)> 1:
  user_and_assistant = parts[1].strip().split("ASSISTANT:")
  user =  user_and_assistant[0].strip()
  assistent =  user_and_assistant[1].strip()


print("Usuario:", user)
print("\nCoy Trainer:",assistent)
```
